```python
# Loarding the model Random Ferest
from joblib import load
import urllib.request, json
import pandas as pd

model = load("weather_decision_model.joblib")
```

```python
# Labels for the model
labels = ["Ice", "Normal", "Rain", "Snow"]

# data from the API at the moment
data = None

# Calling the weather API
with urllib.request.urlopen("https://edu.frostbit.fi/api/road_weather/2025/") as url:
    data = json.load(url)

    
# lets predict using API data
if data != None:
    print(data)
    print()

    # API data to pandas dataframe
    tester_row = pd.DataFrame([data])

    print("All probabilities by category:")
    print(model.predict_proba(tester_row))
    print()

    print("Current weather condition:")
    result = labels[model.predict(tester_row)[0]]
    print(result)
    print("-------------------")

    # Printing the result
    if result == "Normal":
        print("Normal weather, drive with normal speed!")
    elif result == "Rain":
        print("Raining. Reduce speed by 20%")
    elif result == "Snow":
        print("Snowing. Reduce speed by 40%")
    elif result == "Ice":
        print("Icy! Reduce speed by 60%!")
```
